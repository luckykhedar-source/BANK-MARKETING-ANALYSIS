```python
#Standardize Categorical Columns
cat_cols = df.select_dtypes(include=["object"]).columns

for col in cat_cols:
  df[col] = df[col].str.strip().str.lower()
```

```python
#Encode Target Variable (y)
df["y"] = df["y"].map({
    "no":0,
    "yes":1
})
print(df["y"].value_counts())
```
